```python
import sys
project_root = "../.."
sys.path.append(project_root)

import seaborn as sns
import torch
from IPython.display import HTML

# Import project packages
from src.simulation.visualization.animation import create_vector_field_animation_2d
from src.simulation.equations.burgers2d import BurgersEquation2D

device = torch.device("cpu")
sns.set_style("ticks")
sns.set_context("paper")
```

```python
eq = BurgersEquation2D(
    Lx=1.0,
    Ly=1.0,
    T=0.04,
    n_x=64,
    n_y=64,
    n_t=100,
    viscosity={"type": "constant", "value": 0.01},
    ic={"type":"grf", "ell": 0.1, "std_u": 0.1, "std_v": 1.0, "mean_u": 0.0, "mean_v": 0.0, "shared": False, "normalize": True, "seed": 12},

    source=None,
    CFL=0.5,
    second_order=True,
    bc={"type": "Periodic"},  
    advection_scheme="fdm2_euler_fixed",
)

snapshots = eq.solve()
animation = create_vector_field_animation_2d(snapshots,fps=30)
HTML(animation.to_jshtml())
```
